```python
class Flight:
    def __init__(self, flight_id, departure, destination, price, airline):
        self.flight_id = flight_id
        self.departure = departure
        self.destination = destination
        self.price = price
        self.airline = airline


flights = [
    Flight(201, "Karachi", "Dubai", 42000, "Emirates"),
    Flight(202, "Karachi", "Dubai", 38000, "FlyDubai"),
    Flight(203, "Karachi", "Istanbul", 72000, "Turkish Airlines"),
    Flight(204, "Lahore", "Dubai", 47000, "Emirates"),
    Flight(205, "Islamabad", "Dubai", 52000, "Airblue"),
    Flight(206, "Lahore", "Istanbul", 69000, "Turkish Airlines"),
    Flight(207, "Karachi", "Doha", 50000, "Qatar Airways")
]


print("======================================")
print("          SMART TRAVEL AGENT")
print("======================================")

departure = input("Enter departure city: ").strip().title()
destination = input("Enter destination city: ").strip().title()

matched_flights = []

for flight in flights:
    if flight.departure == departure and flight.destination == destination:
        matched_flights.append(flight)


if len(matched_flights) == 0:
    print("\nNo flights found for this route.")

else:
    # Sort flights by price
    matched_flights.sort(key=lambda flight: flight.price)

    print("\n===== AVAILABLE FLIGHTS =====")

    for flight in matched_flights:
        print(
            "Flight ID:", flight.flight_id,
            "| Airline:", flight.airline,
            "| Price: Rs.", flight.price
        )

    print("\nBest Recommended Flight:")
    print(
        "Flight ID:", matched_flights[0].flight_id,
        "| Airline:", matched_flights[0].airline,
        "| Price: Rs.", matched_flights[0].price
    )

    choice = int(input("\nEnter Flight ID to book: "))

    booking_found = False

    for flight in matched_flights:
        if flight.flight_id == choice:
            print("\n======================================")
            print("         BOOKING CONFIRMED")
            print("======================================")
            print("Flight ID:", flight.flight_id)
            print("Airline:", flight.airline)
            print("Departure:", flight.departure)
            print("Destination:", flight.destination)
            print("Price: Rs.", flight.price)

            booking_found = True
            break

    if booking_found == False:
        print("\nInvalid Flight ID.")
```
